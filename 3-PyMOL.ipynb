# MORPH ? Part 3: inspect mapped structures in PyMOL

This walkthrough runs in a local PyMOL session. It uses the helper scripts in `src/3-pymol/` to color mapped PDB files and inspect variant positions.

The mapped PDB B-factor stores either gnomAD frequency bins or AlphaMissense scores. The variant chains are visualization constructs, not experimentally validated mutant proteins.


## 1. Start PyMOL and load the gnomAD helper

Run these commands from the repository root, or replace the paths with absolute paths.

```pymol
run src/3-pymol/color_gnom.py
load data/2-processed/morph_outputs/af/P10635_AF_gnomad.pdb, morph_af_model
```


## 2. Color overall allele frequency

`morph_af` colors the structure from the frequency bins stored in the B-factor field. The default highlights the higher-frequency bins; pass `1` to include a wider range.

```pymol
morph_af sel=morph_af_model
morph_af sel=morph_af_model, 1
```


## 3. Inspect a residue and its alternate side chains

`pos_var` creates selections for the reference and variant chains at one residue. The default shows non-matching variant chains. Use `0` as the third argument to show every chain.

```pymol
pos_var sel=morph_af_model, 123
pos_var sel=morph_af_model, 123, 0
show sticks, morph_af_model_pos_123_*_sc
```


## 4. Color ancestry-frequency output

Category mapping uses one variant chain per selected gnomAD category. Load the category output and run `morph_anc`.

```pymol
load data/2-processed/morph_outputs/categories/P10635_category_gnomad.pdb, morph_category_model
morph_anc sel=morph_category_model
morph_anc sel=morph_category_model, 1
```


## 5. Color AlphaMissense output

AlphaMissense outputs store mean scores in B-factors. Load `color_am.py` and apply `coloram` to the mapped object.

```pymol
run src/3-pymol/color_am.py
load data/2-processed/morph_outputs/am/P10635_AM.pdb, morph_am_model
coloram sel=morph_am_model
```


## Interpretation notes

- gnomAD bins are stored as integer B-factors; bin 42 means no retained frequency annotation at that position.
- AlphaMissense B-factors are mean pathogenicity scores, not gnomAD frequencies.
- Variant chains combine substitutions across positions for visualization and should not be interpreted as observed haplotypes.
- Keep the original PDB if you need its original B-factors or model-confidence values.
