# MORPH Part 3: inspect mapped structures in PyMOL

This walkthrough explains how to inspect MORPH models using the open source PyMOL molecualr structure viewer. 



## 1. Prerequisites: Installing Conda and PyMOL

Before starting this tutorial, make sure you have **Conda** and the **open-source version of PyMOL** installed.

### Step 1a: Install Conda

If you don't already have Conda installed, download and install [Miniconda](https://www.anaconda.com/docs/getting-started/miniconda/install) by following the instructions for your operating system.

### Step 1b: Install PyMOL

Once Conda is installed, open a terminal (or Miniconda Prompt on Windows) and activate your base Conda environment:

```bash
conda activate base
```

Then, install the open-source version of PyMOL from the `conda-forge` channel:

```bash
conda install -c conda-forge pymol-open-source
```

### Step 1c: Launch PyMOL

After installation, you can launch PyMOL directly from your terminal by running:

```bash
pymol
```

You are now ready to proceed with the tutorial!

### Step 3b Visualizing allele frequencies with `morph_af`

Once your MORPH-generated PDB file is loaded in PyMOL, use the following command to visualize gnomAD missense allele frequencies:

```python
morph_af sel=your_AF_object_name
```




### Frequency bins and color scheme



Allele frequencies are encoded as discrete bins in the PDB B-factor field. Residues are colored in **shades of grey** (AF < 10⁻⁴), **red** (10⁻⁴–10⁻³), **dark red** (10⁻³–10⁻²), **blue** (10⁻²–0.51), and **purple** (≥ 0.51). **Pale yellow** indicates positions without retained frequency annotations.

Running `morph_af` creates a **`_variant_spheres`** object highlighting positions that meet the selected allele-frequency cutoff, along with selections for the **reference side chains** (`_maj_residue`) and the **first, second, and third alternative alleles** (`_min_residue`, `_min_residue2`, `_min_residue3`) at positions with multiple missense variants.

Importantly, **residue coloring reflects the frequency of the most frequent alternative allele** at each position, rather than the individual frequencies of the second or third alternatives.

**Note to users:** MORPH-generated PDB files contain multiple overlapping chains representing the reference and alternative amino acids. The `morph_af` command selectively displays the relevant residues and side chains, hiding the others by default. However, manually changing the visualization settings of the entire object (e.g., **Show → Sticks**) may reveal multiple overlapping chains simultaneously.


### Step 3b Visualizing further allele frequencies with `morph_af`

By default, alternative side chains are displayed for bins 5–8 (AF ≥ 10⁻⁴). Use `morph_af sel=your_object_name, 1` to also include bin 4 (AF ≥ 10⁻⁵), highlighted in pink.

### Frequency bins and color scheme

Allele frequencies are encoded as discrete bins in the PDB B-factor field. Unlike `morph_af`, the `morph_anc` command colors variants **by gnomAD population category rather than allele frequency**. The reference protein (chain A) is displayed in **light grey**, while category-specific variants are colored **yellow, orange, red, violet, blue, light blue, or green**, depending on their assigned chain (T–Z). The correspondence between chains and population categories is recorded in the PDB file remarks.

Running `morph_anc` creates a **`_variations`** sphere object for each available category (e.g., `_T_variations`, `_U_variations`), highlighting positions that meet the selected allele-frequency cutoff. Corresponding **alternative side-chain selections** (`_T_min_residue`, `_U_min_residue`, etc.) are also created, alongside the **reference residue selection** (`_maj_residue`).

By default, only variants with **AF ≥ 10⁻³** (bins 6–8) are highlighted. Using `morph_anc sel=your_object_name, 1` extends the cutoff to **AF ≥ 10⁻⁴** (bins 5–8), with the newly included variants displayed in lighter shades of their respective category colors.

Importantly, **each category chain represents the most frequent retained alternative allele at each position within that population category**, rather than the first, second, or third most frequent alternatives across all populations.

**Note to users:** MORPH-generated PDB files contain multiple overlapping chains representing the reference and category-specific alternative amino acids. The `morph_anc` command selectively displays the relevant residues and side chains, hiding the others by default. However, manually changing the visualization settings of the entire object (e.g., **Show → Sticks**) may reveal multiple overlapping chains simultaneously.
